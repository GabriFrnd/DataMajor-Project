```python
import os

import pandas as pd
from pymongo import MongoClient
```

```python
mongo_uri = os.getenv('MONGODB_URI', 'mongodb://localhost:27017')
client = MongoClient(mongo_uri)
database = client['data_major']
collection = database['facebook_reviews']
mongo_uri
```

```python
document_count = collection.count_documents({})
document_count
```

```python
sample_documents = list(collection.find({}, {'_id': 0}).limit(5))
sample_df = pd.DataFrame(sample_documents)
sample_df
```
